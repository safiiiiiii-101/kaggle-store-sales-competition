```python
import pandas as pd
import numpy as np

path = '/kaggle/input/competitions/playground-series-s5e8/'  # verify this path with os.walk if it doesn't work first try

train = pd.read_csv(path + 'train.csv')
test = pd.read_csv(path + 'test.csv')

print(train.shape)
print("--------------------")
print(train.head())
print("--------------------")
print(train.dtypes)
```

```python
cat_cols = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome']

for col in cat_cols:
    test[col] = test[col].astype('category').cat.codes
```

```python
train_raw = pd.read_csv(path + 'train.csv')
test_raw = pd.read_csv(path + 'test.csv')

print(train_raw['marital'].astype('category').cat.categories)
print(test_raw['marital'].astype('category').cat.categories)
```

```python
from sklearn.model_selection import train_test_split

tr,val=train_test_split(train,test_size=0.2,stratify=train['y'],random_state=42)
```

```python
print(tr['y'].value_counts(normalize=True))
print(val['y'].value_counts(normalize=True))
```

```python
features=['age', 'job', 'marital', 'education', 'default', 'balance', 'housing', 'loan', 'contact', 'day', 'month', 'duration', 'campaign', 'pdays', 'previous', 'poutcome']
```

```python
import xgboost as xgb

dtrain=xgb.DMatrix(tr[features],label=tr['y'])
dval=xgb.DMatrix(val[features],label=val['y'])
```

```python
params={
    'objective':'binary:logistic',
    'eval_metric' : 'auc',
    'eta':0.1,
    'max_depth':6
}
model = xgb.train(
    params,
    dtrain,
    num_boost_round=300,
    evals=[(dval, 'val')],
    early_stopping_rounds=30,
    verbose_eval=50
)
```

```python
sample_submission = pd.read_csv(path + 'sample_submission.csv')
print(sample_submission.head())
```

```python
dtest = xgb.DMatrix(test[features])
test_preds = model.predict(dtest)

submission = pd.DataFrame({'id': test['id'], 'y': test_preds})
submission.to_csv('submission.csv', index=False)
print(submission.head())
```

```python
importance = model.get_score(importance_type='gain')
imp_df = pd.DataFrame(importance.items(), columns=['feature','importance']).sort_values('importance', ascending=False)
print(imp_df)
```

```python
features_new=[f for f in features if f!='duration']
dtrain=xgb.DMatrix(tr[features_new],label=tr['y'])
dval=xgb.DMatrix(val[features_new],label=val['y'])
params={
    'objective':'binary:logistic',
    'eval_metric' : 'auc',
    'eta':0.1,
    'max_depth':6
}
model = xgb.train(
    params,
    dtrain,
    num_boost_round=300,
    evals=[(dval, 'val')],
    early_stopping_rounds=30,
    verbose_eval=50
)
```

```python
importance = model.get_score(importance_type='gain')
imp_df = pd.DataFrame(importance.items(), columns=['feature','importance']).sort_values('importance', ascending=False)
print(imp_df)
```

```python
dtest = xgb.DMatrix(test[features_new])
test_preds = model.predict(dtest)

submission = pd.DataFrame({'id': test['id'], 'y': test_preds})
submission.to_csv('submission.csv', index=False)
print(submission.head())
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score

skf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

X=train[features_new]
y=train['y']

auc_scores=[]

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]
    dtrain = xgb.DMatrix(X_tr, label=y_tr)
    dval = xgb.DMatrix(X_val, label=y_val)
    model = xgb.train(
        params,
        dtrain,
        num_boost_round=300,
        evals=[(dval, 'val')],
        early_stopping_rounds=30,
        verbose_eval=False
    )
    val_preds = model.predict(dval)
    fold_auc = roc_auc_score(y_val, val_preds)
    auc_scores.append(fold_auc)
    print(f"Fold {fold} AUC: {fold_auc:.5f}")
print(f"Mean AUC: {np.mean(auc_scores):.5f}")


print(auc_scores)
print(len(auc_scores))
```
